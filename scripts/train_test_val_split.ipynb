# Split bilingual dataset into train, test, validation

The filtered dataset (74,870 sentence pairs) was randomly split into training, validation, and test sets using an 8:1:1 ratio. A fixed random seed was used to ensure reproducibility.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
# load filtered bilingual dataset
df = pd.read_csv("../data/bilingual/original_bilingual_filtered.csv")
```

```python
# keep only necessary columns (nk, sk)
df = df[["nk", "sk"]].dropna()
```

```python
# train (80%) + temp (20%)
train_df, temp_df = train_test_split(df, test_size=0.2, random_state=42)
```

```python
# split temp_df into val (10%) and test (10%)
val_df, test_df = train_test_split(temp_df, test_size=0.5, random_state=42)
```

```python
# save files (TSV format for KoBART code)
train_df.to_csv("../data/bilingual/train.tsv", sep="\t", index=False)
val_df.to_csv("../data/bilingual/val.tsv", sep="\t", index=False)
test_df.to_csv("../data/bilingual/test.tsv", sep="\t", index=False)
```
